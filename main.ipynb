```python
import pandas as pd
data = {"A": [5, 10, 15], "B": [20, 25, 30], "C": [35, 40, 45]}




df = pd . DataFrame(data, index=["X", "Y", "Z"])

df.head()
```

```python
df["D"] = df.apply(lambda row: row.A + row.B, axis=1)
df.head()
```
